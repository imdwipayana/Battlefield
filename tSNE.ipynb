```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.manifold import TSNE
from sklearn.preprocessing import normalize

# ----- 1️⃣ Create synthetic embeddings -----
np.random.seed(42)

# Let's say we have 5 "semantic" clusters
clusters = []
for i in range(5):
    center = np.random.randn(50, 10) + i * 5  # distinct centers
    clusters.append(center)
embeddings = np.vstack(clusters)

# ----- 2️⃣ Normalize embeddings -----
embeddings_norm = normalize(embeddings, axis=1)

# ----- 3️⃣ Run t-SNE -----
tsne = TSNE(n_components=2, random_state=42, perplexity=30)
tsne_orig = tsne.fit_transform(embeddings)
tsne_norm = tsne.fit_transform(embeddings_norm)

# ----- 4️⃣ Visualize -----
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].scatter(tsne_orig[:, 0], tsne_orig[:, 1], c=np.repeat(range(5), 50), cmap='tab10')
axes[0].set_title("t-SNE on raw embeddings (un-normalized)")

axes[1].scatter(tsne_norm[:, 0], tsne_norm[:, 1], c=np.repeat(range(5), 50), cmap='tab10')
axes[1].set_title("t-SNE on normalized embeddings")

plt.show()
```
