### Plot the number of daily delays in 2014

```python
selected_2014 = selected_2014.copy() # make a copy of dataframe
selected_2014['Date'] = pd.to_datetime(selected_2014['Date']) # transform "Date" column to datetime format
daily_counts = selected_2014.groupby(selected_2014['Date'].dt.date).size().reset_index(name='number_delay') # Aggregate by counting the number of delays for each day

# plot
plt.figure(figsize=(12, 6))
plt.plot(daily_counts['Date'], daily_counts['number_delay'], linewidth=1.2)
plt.title('Daily Delays 2014')
plt.xlabel('Date')
plt.ylabel('Number of Delays')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

# 3. Plot the number of delays per day of the week

```python
# Ensure capitalization is consistent (optional)
df['Day'] = df['Day'].str.capitalize()

# Define the correct weekday order
days_order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]

# Count the number of accidents per day
accidents_by_day = df['Day'].value_counts().reindex(days_order)

# Display the count
print(accidents_by_day)
```

```python
plt.figure(figsize=(8, 5))
accidents_by_day.plot(kind='bar', color='teal', edgecolor='black')

plt.title('Number of Delays by Day of the Week (2014–2025)')
plt.xlabel('Day of Week')
plt.ylabel('Number of Delays')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

## Yearly number of delays 2014-2025

```python
import pandas as pd

df['Date'] = pd.to_datetime(df['Date'])

# Extract year from Date
df['Year'] = df['Date'].dt.year

# Count number of accidents per year
yearly_counts = df.groupby('Year').size().reset_index(name='num_accidents')

print(yearly_counts)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))
plt.plot(yearly_counts['Year'], yearly_counts['num_accidents'], marker='o', linewidth=2)
plt.title('Yearly Number of Delays (2014–2025)')
plt.xlabel('Year')
plt.ylabel('Number of Delays')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## Plot the average of delays in minute

```python
import pandas as pd
import matplotlib.pyplot as plt

# Extract year and month
df['YearMonth'] = df['Date'].dt.to_period('M')

# Compute monthly average delay
monthly_avg = df.groupby('YearMonth')['Min Delay'].mean().reset_index()

# Convert back to datetime for plotting
monthly_avg['YearMonth'] = monthly_avg['YearMonth'].dt.to_timestamp()

# Plot
plt.figure(figsize=(12, 6))
plt.plot(monthly_avg['YearMonth'], monthly_avg['Min Delay'], color='steelblue', linewidth=2)
plt.title('Monthly Average Delay (in Minutes)')
plt.xlabel('Month')
plt.ylabel('Average Delay (minutes)')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## Plot the time span of delays

```python
import pandas as pd
import matplotlib.pyplot as plt

# Ensure Time column is in datetime.time format
df['Time'] = pd.to_datetime(df['Time'], format='%H:%M:%S', errors='coerce').dt.time

# Convert time to hours (0–24 float value)
df['hour'] = df['Time'].apply(lambda x: x.hour + x.minute / 60 if pd.notnull(x) else None)

# Define bins and labels for 3-hour intervals
bins = [0, 3, 6, 9, 12, 15, 18, 21, 24]
labels = ['00:00–03:00', '03:00–06:00', '06:00–09:00', '09:00–12:00', 
          '12:00–15:00', '15:00–18:00', '18:00–21:00', '21:00–00:00']

# Categorize each accident into a time bin
df['time_range'] = pd.cut(df['hour'], bins=bins, labels=labels, right=False, include_lowest=True)

# Count the number of accidents in each time range
time_counts = df['time_range'].value_counts().sort_index().reset_index()
time_counts.columns = ['Time Range', 'Number of Delays']

# Plot
plt.figure(figsize=(10, 5))
plt.bar(time_counts['Time Range'], time_counts['Number of Delays'], color='tomato')
plt.title('Number of Delays by 3-Hour Time Interval')
plt.xlabel('Time Range')
plt.ylabel('Number of Delays')
plt.xticks(rotation=45)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

The number of delays usually happened in the peak hours 06:00-09:00
