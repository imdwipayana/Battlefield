```python
import math

def combination(n, r):
    return math.factorial(n) // (math.factorial(r) * math.factorial(n - r))

p = 1/combination(50, 7)
q = 1-p
qq = 1-(q**300)
Ex = 1/qq
result = Ex/104
print(result)
```
