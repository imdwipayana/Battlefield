```python
import requests

url = "https://en.wikipedia.org/w/api.php"
params = {
    "action": "query",
    "format": "json",
    "titles": "Machine learning",
    "prop": "extracts",
    "exintro": True,
    "explaintext": True
}
headers = {
    "User-Agent": "EkaLearningBot/1.0 (https://github.com/ekadwipayana; contact: you@example.com)"
}

response = requests.get(url, params=params, headers=headers)
response.raise_for_status()

data = response.json()
page = next(iter(data["query"]["pages"].values()))

print("Title:", page["title"])
print("\nSummary:\n", page["extract"])
```

```python
import pandas as pd

topics = ["Machine learning", "Deep learning", "Artificial intelligence", "Computer vision"]
results = []

for topic in topics:
    params["titles"] = topic
    r = requests.get(url, params=params, headers=headers)
    r.raise_for_status()
    data = r.json()
    page = next(iter(data["query"]["pages"].values()))
    results.append({"Title": page["title"], "Summary": page["extract"]})

df = pd.DataFrame(results)
print(df)
df.to_csv("wiki_summaries.csv", index=False)
```

```python
url = "https://en.wikipedia.org/wiki/Alan_Turing"
headers = {"User-Agent": "EkaLearningBot/1.0 (https://github.com/ekadwipayana; contact: you@example.com)"}

response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text, "lxml")
```

```python
infobox = soup.find("table", {"class": "infobox"})
```

```python
print(infobox.prettify()[:1000])  # just first 1000 chars for preview
```

```python
data = {}
for row in infobox.find_all("tr"):
    header = row.find("th")
    value = row.find("td")
    if header and value:
        key = header.get_text(" ", strip=True)
        val = value.get_text(" ", strip=True)
        data[key] = val
```

```python
for key, val in data.items():
    print(f"{key}: {val}")
```

```python
import pandas as pd

df = pd.DataFrame(list(data.items()), columns=["Field", "Value"])
df.to_csv("alan_turing_infobox.csv", index=False)
```

```python
def scrape_wikipedia_infobox(title):
    url = f"https://en.wikipedia.org/wiki/{title.replace(' ', '_')}"
    headers = {"User-Agent": "EkaLearningBot/1.0 (https://github.com/ekadwipayana; contact: you@example.com)"}
    
    response = requests.get(url, headers=headers)
    soup = BeautifulSoup(response.text, "lxml")
    infobox = soup.find("table", {"class": "infobox"})
    tables = soup.find_all("table", {"class": "wikitable"})
    
    if not infobox:
        return None  # no infobox found
    
    data = {}
    for row in infobox.find_all("tr"):
        header = row.find("th")
        value = row.find("td")
        if header and value:
            key = header.get_text(" ", strip=True)
            val = value.get_text(" ", strip=True)
            data[key] = val
    
    return data
```
