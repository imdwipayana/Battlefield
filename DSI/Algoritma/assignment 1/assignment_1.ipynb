````python
from typing import List

def first_duplicate(nums: List[int]) -> int:
    flag = set() # Create flag to collect a unique member of list
    for num in nums:
        if num in flag: # looking whether the list's member already in the flag?
            return num # return the first number of the list that already in the flag
        flag.add(num) # add the number into flag if the number is not in the flag
    return -1

nums_1 = [3, 1, 4, 2, 5, 1, 6]
print("The first duplicate from num_1 is ", first_duplicate(nums_1))

nums_2 = [7, 8, 9, 10]
print("The first duplicate from num_2 is ",first_duplicate(nums_2))

nums_3 = [4, 5, 6, 4, 6]
print("The first duplicate from num_3 is ",first_duplicate(nums_3))
````

````python
def is_valid_brackets(s: str) -> bool:
    stack = []
    pairs = {')': '(', ']': '[', '}': '{'}
    
    for ch in s:
        if ch in pairs.values():  
            stack.append(ch)
        elif ch in pairs:  
            if not stack or stack[-1] != pairs[ch]: # Comparing the list's members
                return False
            stack.pop()
        else:
            return False  
    
    return not stack

s_1 = "([]{})"
print("Is s_1 a valid bracket sequence? The answer is: ", is_valid_brackets(s_1))

s_2 = "([)]"
print("Is s_2 a valid bracket sequence? The answer is: ",is_valid_brackets(s_2))

s_3 = "()[]{}"
print("Is s_3 a valid bracket sequence? The answer is: ",is_valid_brackets(s_3))

s_4 = "[{]}"
print("Is s_4 a valid bracket sequence? The answer is: ",is_valid_brackets(s_4))
````

````python
from typing import List

def move_zeros_to_end(nums: List[int]) -> List[int]:
    non_zero_index = 0
    
    # Move non zero elements forward
    for i in range(len(nums)):
        if nums[i] != 0:
            nums[non_zero_index] = nums[i]
            non_zero_index += 1
    
    # Move all the zeros at back of the list
    for i in range(non_zero_index, len(nums)):
        nums[i] = 0
    
    return nums

nums_11 = [0, 1, 0, 3, 12]
print("After move all the zero at the end, the num_11 list will be ", move_zeros_to_end(nums_11))

nums_22 = [4, 0, 5, 0, 0, 6]
print("After move all the zero at the end, the num_22 list will be ", move_zeros_to_end(nums_22))
````
