```python
import hashlib

def hash_to_range(input_string: str) -> int:
     hash_object = hashlib.sha256(input_string.encode())
     hash_int = int(hash_object.hexdigest(), 16)
     return (hash_int % 3) + 1
input_string = "i made eka"
result = hash_to_range(input_string)
print(result)
```

# Question Three: Missing Number in Range
 
  You are given a list containing `n` integers in the range `[0, n]`. Return a list of numbers that are missing from the range `[0, n]` of the array. If there is no missing number, return -1. Note, all the integers in the list may not be unique.
  
  ## Examples

  ### Example 1

  Input: `lst = [0, 2]`

  Output: [1]

  ### Example 2

  Input: `lst = [5, 0, 1]`

  Output: [2, 3, 4]

  ### Example 3

  Input: `lst = [6, 8, 2, 3, 5, 7, 0, 1, 10]`

  Output: [4, 9]

```python
# This is for n is non negative
from typing import List, Union

def missing_num(nums: List[int]) -> Union[List[int], int]:
    max_value = max(nums) # Find the max value of the list members
    full_range = set(range(0, max_value + 1)) # Create a new list that range [0,max_value]
    missing = sorted(list(full_range - set(nums))) # Remove the full_range list that contains nums list values

    if len(missing) == 0: # Check the length of missing list
        return -1 # If there is no missing number, return -1
    return missing # If the length of missing list != 0, then return the missing list


nums_1 = [0, 2]
print(missing_num(nums_1))

nums_2 = [5, 0, 1]
print(missing_num(nums_2))

nums_3 = [6, 8, 2, 3, 5, 7, 0, 1, 10]
print(missing_num(nums_3))

nums_4 = [0,1,2,2,5] # it can handle the not unique list values.
print(missing_num(nums_4))

nums_5 = [0]
print(missing_num(nums_5))

nums_6 = [0,-5] # Use the next function where list in the range of  [max_value and min_value]
print(missing_num(nums_6))
```

```python
# This is for the list in range of [min_value, max_value]
from typing import List, Union

def missing_num(nums: List[int]) -> Union[List[int], int]:
    max_value = max(nums) # Find the max value of the list members
    min_value = min(nums)
    full_range = set(range(min_value, max_value + 1)) # Create a new list that range [0,max_value]
    missing = sorted(list(full_range - set(nums))) # Remove the full_range list that contains nums list values

    if len(missing) == 0: # Check the length of missing list
        return -1 # If there is no missing number, return -1
    return missing # If the length of missing list != 0, then return the missing list

nums_6 = [0,-5] 
print(missing_num(nums_6))

nums_7 = [4,4,10,10,-5,-5] 
print(missing_num(nums_7))
```

```python
# Your answer here
def is_valid_brackets(s: str) -> bool:
    empty_list = []
    defined_brackets = {"(": ")", "[": "]", "{": "}"}

    for element in s:
        if element in defined_brackets.keys(): # Reference: https://developers.google.com/edu/python/dict-files
            empty_list.append(element)

        elif element in defined_brackets.values(): # Reference: https://developers.google.com/edu/python/dict-files
            if not empty_list:
                return False
            last_opening_bracket = empty_list.pop(-1) # Reference: https://www.programiz.com/python-programming/methods/list/pop
            if defined_brackets[last_opening_bracket] != element:
                return False

        else:
            return False

    print(f"Check spot for empty_list: {empty_list}") # CYC: Add a check for empty_list: if it's True, the output should be []; otherwise, the output should be something other than an empty list. 
    return empty_list == []
```
