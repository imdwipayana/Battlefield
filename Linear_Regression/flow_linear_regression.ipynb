Refined Steps for Linear Regression Workflow

1. Data preprocessing
   * Handle missing values (imputation or removal).
   * Encode categorical variables properly (e.g., one-hot encoding, dummy variables).
   * Standardize/normalize numerical features if needed (especially for interpretability in regression).
2. Check multicollinearity
   * Use VIF (Variance Inflation Factor) for numerical predictors.
   * For categorical vs categorical: Cramér’s V or chi-square test.
   * For categorical vs numerical: ANOVA F-test or correlation ratio (η²).
   * Decide which features to remove or combine if high collinearity is detected.
3. Build initial feature set
   * Keep all remaining features after screening.
5. Fit the model
   * Train the regression model on your dataset.
7. Evaluate model fit
   * Use RMSE, MAE, R², adjusted R² for performance.
   * Consider cross-validation instead of just train-test split to check robustness.
9. Statistical inference (hypothesis testing on coefficients)
    * Look at p-values and confidence intervals for coefficients.
    * Remove features with high p-values (not statistically significant), but do this carefully:
      a. Some features may not be significant alone but still improve overall model (consider adjusted R² or AIC/BIC).
11. Iterative refinement
    * Refit after removing insignificant or redundant predictors.
    * Re-check VIF after each round.
    * Balance parsimony (simpler model) vs explanatory power.
12. Final model validation
    * Check residuals (should be approximately normally distributed, homoscedastic, no strong autocorrelation).
    * Ensure assumptions of linear regression hold.
    
⚖️ Key insights:

Your step (6) "Check hypothesis testing" and step (7) "Remove not significant features" are good, but always combine with adjusted R², AIC, or cross-validation error — p-values alone can be misleading.

If features are collinear, hypothesis tests become unreliable (that’s why multicollinearity check comes before inference).

Iteration is expected — regression is rarely a one-shot fit.

🔹 How to handle high collinearity

1. Drop features iteratively
   * Remove the feature with the highest VIF.
   * Refit regression.
   * Recalculate VIFs.
   * Repeat until all VIFs are below a threshold (commonly < 5 or < 10).
2. Combine features
   * If two features are highly correlated (e.g., Height_cm and Height_inch), keep only one or combine them into a new feature.
   * Another approach is dimension reduction:
     a. PCA (Principal Component Analysis): transform correlated variables into uncorrelated components, then run regression on those.
     b. Partial Least Squares (PLS) regression: designed for highly collinear predictors.
3. Regularization instead of feature removal
   * Ridge regression (L2 penalty) shrinks coefficients of correlated predictors but keeps them.
   * Lasso regression (L1 penalty) can shrink some coefficients exactly to zero → automatic feature selection.
   * ElasticNet (L1 + L2) balances both.
