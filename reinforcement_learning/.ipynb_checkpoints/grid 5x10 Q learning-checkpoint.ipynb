## 2. Environment Setup

```python
# Grid size
ROWS = 6
COLS = 10

# Rewards grid (everything 0 except special cells)
rewards = np.zeros((ROWS, COLS))

# Set special grid rewards
rewards[1, 7] = 5
rewards[4, 2] = 3
rewards[2, 9] = -2

# Starting position (row, col)
start_state = (3, 3)

# Actions: Up, Down, Left, Right
actions = {
    0: (-1, 0),  # Up
    1: (1, 0),   # Down
    2: (0, -1),  # Left
    3: (0, 1)    # Right
}

def is_terminal(state):
    r, c = state
    return rewards[r, c] != 0
```
