```python
import numpy as np
import pandas as pd

np.random.seed(42)

# Config
N_CUSTOMERS = 5000
N_DAYS = 180

# Customer segments
segments = ["loyal", "average", "at_risk", "churn_prone"]
segment_probs = [0.20, 0.40, 0.25, 0.15]
segment_assignment = np.random.choice(segments, N_CUSTOMERS, p=segment_probs)

# Base purchase probabilities
segment_purchase_prob = {
    "loyal": 0.20,
    "average": 0.10,
    "at_risk": 0.05,
    "churn_prone": 0.02
}

# Marketing actions
ACTIONS = ["no_action", "email", "small_coupon", "big_coupon"]
action_effects = {
    "no_action": 0.00,
    "email": 0.03,
    "small_coupon": 0.07,
    "big_coupon": 0.15
}

rows = []

for cust_id in range(N_CUSTOMERS):
    seg = segment_assignment[cust_id]
    base_prob = segment_purchase_prob[seg]

    for day in range(N_DAYS):
        # choose an action (random for now)
        action = np.random.choice(ACTIONS)

        # probability of purchase today
        p = base_prob + action_effects[action]
        p = min(max(p, 0), 1)

        purchased = np.random.rand() < p
        amount = np.random.exponential(50) if purchased else 0

        rows.append({
            "customer_id": cust_id,
            "day": day,
            "segment": seg,
            "action": action,
            "purchase": int(purchased),
            "amount": amount
        })

df = pd.DataFrame(rows)
df.head()
```
