```python
import pandas as pd
import numpy as np
from sklearn.feature_selection import mutual_info_classif
import scipy.stats as ss

# Example data
df = pd.DataFrame({
    "Color": ["Red", "Blue", "Green", "Blue", "Red", "Green", "Red"],
    "Target": ["Yes", "No", "No", "Yes", "Yes", "No", "Yes"]
})

# --- Chi-square and Cramer's V ---
contingency = pd.crosstab(df["Color"], df["Target"])
chi2, p, dof, ex = ss.chi2_contingency(contingency)

# Cramer's V
n = contingency.sum().sum()
cramers_v = np.sqrt(chi2 / (n * (min(contingency.shape)-1)))

print("Chi-square:", chi2, "p-value:", p)
print("Cramer's V:", cramers_v)

# --- Mutual Information ---
X = pd.get_dummies(df[["Color"]])   # one-hot encoding
y = df["Target"].astype("category").cat.codes  # encode target
mi = mutual_info_classif(X, y, discrete_features=True)
print("Mutual Information:", mi)
```
