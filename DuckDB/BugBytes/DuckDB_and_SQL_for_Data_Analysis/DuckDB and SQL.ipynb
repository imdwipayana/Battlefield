```python
duckdb.read_csv("employee.csv")
```

```python
duckdb.sql("""
SELECT Department, AVG(Salary)
FROM 'employee.csv'
WHERE Name is NOT NULL
GROUP BY Department
ORDER BY AVG(Salary)
""")
```

```python
duckdb.sql("CREATE TABLE employees AS SELECT * FROM 'employee.csv'")
```

```python
duckdb.sql("""
SELECT Department, AVG(Salary)
FROM employees
WHERE Name is NOT NULL
GROUP BY Department
ORDER BY AVG(Salary)
""")
```
