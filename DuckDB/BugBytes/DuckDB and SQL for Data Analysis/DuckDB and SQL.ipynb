```python
duckdb.sql("SELECT * FROM 'employee.csv'")
```

```python
duckdb.read_csv("employee.csv")
```

```python
con = duckdb.connect()
con.execute("""
    CREATE TABLE employees AS 
    SELECT * FROM 'employee.csv'
""")
```

```python
duckdb.sql("CREATE TABLE employees AS SELECT * FROM 'employees.csv'")
```

```python
duckdb.sql("""
SELECT Team, AVG(Salary)
FROM employee
WHERE Team is NOT NULL
GROUP BY Team
ORDER BY AVG(Salary)
""")
```
