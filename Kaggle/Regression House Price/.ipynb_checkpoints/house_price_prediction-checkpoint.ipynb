```python
import pandas as pd

# Show all columns
pd.set_option("display.max_columns", None)
```

```python
house = pd.read_csv("./data/train.csv")
```

```python
house.head()
```

```python
house.info()
```

```python
# Drop column that empty rows more than 5%

# Set threshold (5%)
threshold = 0.05

# Find ratio of missing values per column
missing_ratio = house.isnull().mean()

# Drop columns where missing % > threshold
house_cleaned = house.drop(columns=missing_ratio[missing_ratio > threshold].index)

print("Dropped columns:", missing_ratio[missing_ratio > threshold].index.tolist())
house_cleaned.head()
```

```python
house_cleaned.info()
```

```python
# Do the imputation for missing data where the numerical columns will be imputed by median and the categorical columns will be imputed with the most frequent

from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Identify numerical and categorical columns
num_cols = house_cleaned.select_dtypes(include=["number"]).columns
cat_cols = house_cleaned.select_dtypes(exclude=["number"]).columns

# Define imputers
num_imputer = SimpleImputer(strategy="median").set_output(transform="pandas")
cat_imputer = SimpleImputer(strategy="most_frequent").set_output(transform="pandas")

# Build column transformer
preprocessor_imputer = ColumnTransformer(
    transformers=[
        ("num", num_imputer, num_cols),
        ("cat", cat_imputer, cat_cols)
    ]
).set_output(transform="pandas")

# Create pipeline (optional, useful for ML tasks)
pipeline_imputer = Pipeline(steps=[("preprocessor", preprocessor_imputer)]).set_output(transform="pandas")

# Fit and transform
house_imputed = pipeline_imputer.fit_transform(house_cleaned)

# Convert back to DataFrame
house_imputed = pd.DataFrame(house_imputed, columns=list(num_cols) + list(cat_cols))
```

```python
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Example dataframe
df = pd.DataFrame({
    "Age": [25, 30, None, 40, None],
    "Salary": [50000, None, 60000, 55000, None],
    "Gender": ["Male", None, "Female", "Male", None],
    "City": ["NY", "LA", None, "NY", "SF"]
})

# Identify numerical and categorical columns
num_cols = df.select_dtypes(include=["number"]).columns
cat_cols = df.select_dtypes(exclude=["number"]).columns

# Define imputers
num_imputer = SimpleImputer(strategy="median").set_output(transform="pandas")
cat_imputer = SimpleImputer(strategy="most_frequent").set_output(transform="pandas")

# Build column transformer
preprocessor = ColumnTransformer(
    transformers=[
        ("num", num_imputer, num_cols),
        ("cat", cat_imputer, cat_cols)
    ]
).set_output(transform="pandas")

# Create pipeline
pipeline = Pipeline(steps=[("preprocessor", preprocessor)]).set_output(transform="pandas")

# Fit and transform
df_imputed = pipeline.fit_transform(df)

print(df_imputed.dtypes)
print(df_imputed)
```
